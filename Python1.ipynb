```python
# Set operations
python_students = {"Amit","Rani","Devi","Rajesh"}
java_students = {"Rajesh","Devi","Priya","Neha"}
result1 = python_students.union(java_students)
result2 = python_students.intersection(java_students)
print(result1)
print(result2)
```

```python
#Symmetric difference
python_students = {"Amit","Rani","Devi","Rajesh"}
java_students = {"Rajesh","Devi","Priya","Neha"}
result1 = python_students.symmetric_difference(java_students)
print(result1)
```
